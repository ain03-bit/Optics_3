```python

def compute_shift(ti, tref):
    corr = np.correlate(ti, tref, mode='full')
    shift = np.argmax(corr) - (len(ti) - 1)
    return shift

def shift_trace(trace, shift):
    return np.roll(trace, shift)
```

```python
def baseline_and_smooth(i):
    t = Time[i].values 
    y = CH1[i].values 
    f = CH2[i].values

    # baseline 제거
    n = len(y)
    edge = n // 20
    mask = np.r_[np.arange(edge), np.arange(n-edge, n)]

    coef = np.polyfit(t[mask], y[mask], deg=1)
    baseline = np.polyval(coef, t)
    y_corr = y - baseline

    # smoothing
    win = 7
    if win % 2 == 0:
        win += 1

    #y_smooth = savgol_filter(y_corr, win, polyorder=3)
    f_smooth = savgol_filter(f, win, polyorder=3)
    
    return t, y_corr, f, f_smooth

def find_min_in_range(t, y_corr, f_smooth, f, t_min, t_max):
    # 구간 마스크
    mask = (t >= t_min) & (t <= t_max)

    t_sel = t[mask]
    y_sel = y_corr[mask]
    f_sel = f_smooth[mask]
    #ch1_sel = y[mask] #original
    ch2_sel = f[mask] #original
    # 구간에 데이터가 없을 때
    if len(y_sel) == 0:
        print("⚠ 선택한 구간에 데이터가 없습니다.")
        return None, None

    # 최대값 찾기
    idx = np.argmin(y_sel)
    x_min = t_sel[idx]
    y_min = y_sel[idx]
    f_min = f_sel[idx]
    #ch1_min = ch1_sel[idx]
    ch2_min = ch2_sel[idx]
    return x_min, y_min, f_min, ch2_min


t0, y0, f0, f0_s = baseline_and_smooth(46)
ref = y0 #정제된거에 맞춰서

```

```python
from scipy.signal import savgol_filter


def plot_smoothed_signal(i, ref_trace=None):

    t = Time[i]
    y = CH1[i]
    f = CH2[i]

    # ----------------------------
    # Baseline 제거
    # ----------------------------
    n = len(y)
    edge = n // 20
    mask = np.r_[np.arange(edge), np.arange(n-edge, n)]

    coef = np.polyfit(t[mask], y[mask], deg=1)
    baseline = np.polyval(coef, t)
    y_corr = y - baseline

    # ----------------------------
    # Smoothing
    # ----------------------------
    win = 21
    y_smooth = savgol_filter(y_corr, win, polyorder=3)
    f_smooth = savgol_filter(f, win, polyorder=3)

    # ----------------------------
    # Alignment
    # ----------------------------
    if ref_trace is not None:
        shift = compute_shift(y_smooth, ref_trace)

        y_smooth = np.roll(y_smooth, shift)
        f_smooth = np.roll(f_smooth, shift)
        t        = np.roll(t, shift)
        y        = np.roll(y, shift)
        f        = np.roll(f, shift)

    # ----------------------------
    # Plot
    # ----------------------------
    colour = plt.cm.tab20(random.random())
    plt.plot(t, y_smooth, color=colour)

    if f"data{i}_T" in globals() and f"data{i}_1" in globals():
        plt.scatter(
            globals()[f"data{i}_T"],
            globals()[f"data{i}_1"],
            c='red',
            s=60
        )
    #plt.title(f"Baseline removal + smoothing - {i}")
    #plt.xlabel("Time (s)")
    #plt.ylabel("Signal (CH1)")
    #plt.grid()
    #plt.legend()
    #plt.show()

# 예시 실행
for i in range(0,8):
    #path = f"C:/Users/ainch/OneDrive - Durham University/OPTIC_3/data/tek000{i}.xlsx"
    plot_smoothed_signal(i)
```

```python
def plot_smoothed_with_peaks3(i, ref_trace=None):
    window_length = 11
    polyorder = 3
    prom_scale = 0.2
    dist_scale = 250

    # ----------------------------
    # 1) Load data
    # ----------------------------
    t = Time[i].values
    y = CH1[i].values
    f = CH2[i].values

    # ----------------------------
    # 2) Baseline removal
    # ----------------------------
    n = len(y)
    edge = n // 20
    mask = np.r_[np.arange(edge), np.arange(n-edge, n)]

    coef = np.polyfit(t[mask], y[mask], deg=1)
    baseline = np.polyval(coef, t)
    y_corr = y - baseline

    # ----------------------------
    # 3) Smoothing
    # ----------------------------
    if window_length % 2 == 0:
        window_length += 1

    y_s = savgol_filter(y_corr, window_length, polyorder)
    f_s = savgol_filter(f, window_length, polyorder)

    # ----------------------------
    # 4) Alignment (정렬)
    # ----------------------------
    if ref_trace is not None:
        shift = compute_shift(y_s, ref_trace)

        # shift 모든 신호
        y_s = np.roll(y_s, shift)
        f_s = np.roll(f_s, shift)
        t   = np.roll(t, shift)
        #y   = np.roll(y, shift)
        f   = np.roll(f, shift)

    # ----------------------------
    # 5) Peak detection (정렬된 y_s 기준)
    # ----------------------------
    prom = np.std(y_s) * prom_scale
    dist = len(y_s) // dist_scale

    peaks, _ = find_peaks(-y_s, prominence=prom, distance=dist)

    # ----------------------------
    # 6) Save numeric result (정렬된 t, y_s 기준)
    # ----------------------------
    globals()[f"data{i}_T"] = [float(t[p]) for p in peaks]
    globals()[f"data{i}_1"] = [float(y_s[p]) for p in peaks]
    #globals()[f"data{i}_y"] = [float(y[p]) for p in peaks]
    globals()[f"data{i}_2"] = [float(f_s[p]) for p in peaks]
    globals()[f"data{i}_f"] = [float(f[p]) for p in peaks]

    # ----------------------------
    # 7) Plot
    # ----------------------------
    #plt.figure(figsize=(3,1))
    plt.plot(t, y, color='gray', alpha=0.3)
    plt.plot(t, baseline, color='green', alpha=0.6)
    plt.plot(t, y_s, color='blue')
    plt.scatter(t[peaks], y_s[peaks], c='red', s=60)
for i in range(44,67):
    plot_smoothed_with_peaks3(i)
```
